```python
import pandas as pd
```

```python
df=pd.read_csv("/content/FY_2026_Hospital_Readmissions_Reduction_Program_Hospital.csv")
```

```python
df.head(5000) # missing values are in number of dischargers, footnote, excess readmission ratio, predicted & expected readmission rate, number of readmissions.
```

```python
df = df.drop(columns=["Footnote", "Predicted Readmission Rate", "Expected Readmission Rate", "Start Date", "End Date"]) #You dropped the columns because they added no value to the analysis, and the rows because hospitals didn't have enough patients to report data for those conditions, making them unusable.
```

```python
df = df.dropna(subset=["Number of Discharges", "Excess Readmission Ratio", "Number of Readmissions"])
```

```python
df.head(700)
```

```python
df["Number of Readmissions"] = pd.to_numeric(df["Number of Readmissions"]) #Had to change Number of Readmissions to numeric so it can be used in calculations and analysis
```

```python
df.dtypes
```

```python
df2=pd.read_csv("/content/HCAHPS-Hospital.csv", low_memory=False)
```

```python
df2.head(5000) #get rid of answer percent footnote, linear mean value, rate percent footnote, start and end date, telephone number,
```

```python
df2 = df2.drop(columns=["Address", "City/Town", "ZIP Code", "County/Parish", "Telephone Number", "Patient Survey Star Rating Footnote", "HCAHPS Answer Percent Footnote", "HCAHPS Linear Mean Value", "Number of Completed Surveys Footnote", "HCAHPS Answer Percent", "Survey Response Rate Percent Footnote", "Start Date", "End Date"])
```

```python
df2.head(500)
```

```python
df2 = df2[~df2["Patient Survey Star Rating"].isin(["Not Applicable", "Not Available"])]
```

```python
df2.shape
```

```python
df2["Patient Survey Star Rating"] = pd.to_numeric(df2["Patient Survey Star Rating"], errors='coerce')
df2["Number of Completed Surveys"] = pd.to_numeric(df2["Number of Completed Surveys"], errors='coerce')
df2["Survey Response Rate Percent"] = pd.to_numeric(df2["Survey Response Rate Percent"], errors='coerce') #Had to change Patient Survey Star Rating to numeric so it can be used in calculations and analysis
df2["Facility ID"] = pd.to_numeric(df2["Facility ID"], errors='coerce')
```

```python
df_merged = pd.merge(df, right=df2, left_on="Facility ID", right_on="Facility ID", how = "left")
```

```python
df_merged.shape
```

```python
df_merged = df_merged.rename(columns={"Facility Name_x": "Facility Name"})
df_merged = df_merged.rename(columns={"State_x": "State"})
```

```python
print(df_merged)
print(df_merged.isnull().sum())
```

```python
df_merged = df_merged.drop(columns=["Facility Name_y", "State_y"])
print(df_merged.shape)
```

```python
df_merged= df_merged.dropna(subset=["HCAHPS Measure ID", "HCAHPS Question", "HCAHPS Answer Description", "Patient Survey Star Rating", "Number of Completed Surveys", "Survey Response Rate Percent"])
```

```python
print(df_merged.shape)
```

```python
df_merged.isnull().sum()
```

```python
import sqlite3
import pandas as pd

conn= sqlite3.connect("hospital_data.db")

df_merged.to_sql("hospitals", conn, if_exists="replace", index=False)

print("Database created successfully!")
```

```python
query = """
SELECT State, AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM hospitals
GROUP BY State
ORDER BY avg_excess_readmission_ratio DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT State, AVG("Patient Survey Star Rating") AS avg_patient_survey_rating
FROM hospitals
GROUP BY State
ORDER BY avg_patient_survey_rating DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT "Measure Name", AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM hospitals
GROUP BY "Measure Name"
ORDER BY avg_excess_readmission_ratio DESC
"""

result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT "Facility Name", State, AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM hospitals
GROUP BY "Facility Name"
ORDER BY avg_excess_readmission_ratio DESC
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
result
```
