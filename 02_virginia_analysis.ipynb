```python
import pandas as pd
```

```python
df= pd.read_csv("/content/cleaned_hospital_data.csv")
```

```python
df_virginia = df[df["State"] == "VA"] #Flitering down to virginia
print(df_virginia.shape)
```

```python
print(df_virginia["Facility Name"].nunique()) #going to see how many unique hospitals are in the Virginia dataset
```

```python
df.head()
```

```python
import sqlite3

conn_va = sqlite3.connect("virginia_hospitals.db")
df_virginia.to_sql("virginia", conn_va, if_exists="replace", index=False)

print("Virginia database created!")
```

```python
query = """
SELECT "Facility Name", AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM virginia
GROUP BY "Facility Name"
ORDER BY avg_excess_readmission_ratio DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn_va)
result

#the top 10 worst performing hospitals.
```

```python
query = """
SELECT "Facility Name", AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM virginia
GROUP BY "Facility Name"
ORDER BY avg_excess_readmission_ratio ASC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn_va)
result

#the top 10 best performing hospitals.
```

```python
print(df_virginia["Patient Survey Star Rating"].isnull().sum())
print(df_virginia["Patient Survey Star Rating"].unique())
```

```python
query = """
SELECT "Facility Name", "HCAHPS Question", AVG ("Patient Survey Star Rating") AS avg_rating
FROM virginia
GROUP BY "Facility Name", "HCAHPS Question"
ORDER BY avg_rating ASC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn_va)
result

#patient satisfaction trends at specific Virginia hospitals from worst 10
```

```python
query = """
SELECT "Facility Name", "HCAHPS Question", AVG ("Patient Survey Star Rating") AS avg_rating
FROM virginia
GROUP BY "Facility Name", "HCAHPS Question"
ORDER BY avg_rating DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn_va)
result

#patient satisfaction trends at specific Virginia hospitals from top 10
```

```python
query = """
SELECT "Measure Name", AVG("Excess Readmission Ratio") AS avg_excess_readmission_ratio
FROM virginia
GROUP BY "Measure Name"
ORDER BY avg_excess_readmission_ratio DESC
"""

result = pd.read_sql_query(query, conn_va)
result

#which conditions perform best and worst across Virginia hospitals
```
